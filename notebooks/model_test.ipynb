```python
embedding_model = OllamaEmbeddings(model="paraphrase-multilingual:278m")

sentences = [
    "I am a happy person.",
    "I am a joyful person.",
    "I am a pessimistic person.",
    "I am not an optimistic person."
]
```

```python
doc_embeddings = embedding_model.embed_documents(sentences)
```

```python
print(len(doc_embeddings[0]))
print(f"Document Embedding: {doc_embeddings[0][:5]}")
```

```python
def cosine_similarity(vec1, vec2):
    norm_1 = np.linalg.norm(vec1)
    norm_2 = np.linalg.norm(vec1)
    dot_prod = np.dot(vec1,vec2)
    # print(norm_1, norm_2, dot_prod)
    return round(dot_prod/(norm_1*norm_2),2)
```

```python
print(cosine_similarity(doc_embeddings[0], doc_embeddings[3]))
print(cosine_similarity(doc_embeddings[0], doc_embeddings[1]))
```

```python
rand1 = np.random.randint(0, 768)
rand2 = np.random.randint(0, 768)

if rand1 > rand2:
    rand1, rand2 = rand2, rand1

print(rand1, rand2)
```

```python
print(cosine_similarity(doc_embeddings[0][rand1:rand2], doc_embeddings[3][rand1:rand2]))
print(cosine_similarity(doc_embeddings[0][rand1:rand2], doc_embeddings[1][rand1:rand2]))
```

```python
query = "ল্যাংচেইন বড় ভাষা মডেল দিয়ে অ্যাপ তৈরি সহজ করে।"
docs = [
    "ল্যাংচেইন ব্যবহার করলে বড় ভাষা মডেল ভিত্তিক অ্যাপ বানানো সহজ হয়।",
    "ল্যাংচেইন বড় ভাষা মডেল ব্যবহার করে অ্যাপ ডেভেলপমেন্টকে সহজ করে তোলে।",
    "ল্যাংচেইন ডেটা বিশ্লেষণের জন্য একটি টুল নয়।",
    "বাংলা সাহিত্য শেখার জন্য ল্যাংচেইন ব্যবহার করা যায় না।",
    "ল্যাংচেইন মূলত এআই অ্যাপ্লিকেশন তৈরিতে ব্যবহৃত হয়।"
]

query_emb = embedding_model.embed_query(query)
doc_embs = embedding_model.embed_documents(docs)
```

```python
for i in range(5):
    print(cosine_similarity(query_emb, doc_embs[i]))
#  cosine_similarity(query_emb, doc_embs[0])
```
